```python
import pandas as pd
from os import path, listdir
import gzip
```

```python
pipi = r'L:\Auditdata\GM-Bioinformatics\FOB\data_matrices_for_bloodspot\bo_requested_2026-01'
```

```python
def process_stuff(fool_path):
    print(fool_path)
    dada = []
    with gzip.open(fool_path) as inf:
        longest_line = ""
        longest_line_length = 0
        for line in inf:
            ln_str = line.decode()
            if ln_str.startswith('INSERT INTO'):
                this_len = len(ln_str)
                if this_len > longest_line_length:
                    longest_line_length = this_len
                    longest_line = ln_str
                vals_str = ln_str.split('VALUES')[-1]
                for row_str in vals_str.strip().split('),('):
                    this_entry = []
                    this_element = ""
                    in_string = False
                    for c in row_str:
                        if c == "'":
                            in_string = not in_string
                            continue
                        if in_string:
                            this_element += c
                        else:
                            if c == ",":
                                this_entry.append(this_element)
                                this_element = ""
                    this_dada = {
                        'name': this_entry[1],
                        'probe': this_entry[2],
                        'col': this_entry[3],
                    }
                    for valind, val in enumerate(this_entry[4].split()):
                        try:
                            this_dada[f"p{valind:04d}"] = float(val.strip("'"))
                        except Exception as e:
                            print(row_str, this_entry[4])
                            raise ValueError(e)
                    dada.append(this_dada)
        print( longest_line_length, longest_line)
    return dada

                    


data = []
for popo in listdir(pipi):
    this_dada = pd.DataFrame(process_stuff(path.join(pipi,popo)))
    data.append(this_dada)
    this_dada.to_csv(path.basename(popo)+'_bo.tsv',sep='\t',index=False)
    
```
