```python
import gzip
from os import scandir, path
```

```python
dump_dir_path = "s:/GMling/users/mhg/bloodspot/dump_split"
```

```python
tuple_list = []
for dump_file in scandir(dump_dir_path):
    print (dump_file)
    dump_file_path = path.join(dump_dir_path, dump_file)
    if dump_file_path[-3:] != '.gz':
        continue
    if 'corr' in dump_file_path:
        continue
    if 'alias' in dump_file_path:
        continue
    if 'gene' in dump_file_path:
        continue
    with gzip.open(dump_file_path) as inf:
        for byte_str in inf:
            line = byte_str.decode('utf-8').strip()
            if "VALUES (" in line:
                values = line.split('VALUES (')[1].split(',(')
                for val_row in values:
                    row_spl = val_row.split(',')
                    if len(row_spl) < 6:
                        continue
                    tuple_list.append((row_spl[1].strip().strip("'").strip().upper(), row_spl[2].strip().strip("'").strip().upper(),))
tuple_set = set(tuple_list)
```

```python
with open('all_dem_keys.csv', 'w+') as outf:
    outf.write("gene_name,probe_name\n")
    for gene_name, probe_name in sorted(tuple_set):
        outf.write(f"{gene_name},{probe_name}\n")
```
